```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("done")
```

```python
df = pd.read_csv('data/netflix.csv', encoding='latin1')
df.head()
```

```python
sns.countplot(x='type', data=df)
plt.title('Movies vs TV Shows')

plt.savefig('images/movies_vs_tv.png')
plt.show()
```

```python
df['country'].value_counts().head(10).plot(kind='bar')
plt.title('Top 10 Countries with Most Content')

plt.savefig('images/top_countries.png')
plt.show()
```

```python
### Insight:
USA has the highest number of content on Netflix, followed by other countries.
```

```python
df['release_year'].value_counts().sort_index().plot()
plt.title('Content Released Over Years')

plt.savefig('images/content_year.png')
plt.show()
```
